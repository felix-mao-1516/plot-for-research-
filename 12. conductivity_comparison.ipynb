```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

# =========================
# 1. Data
# =========================
data = {
    "E5-G9": {"x": 0.80, "y": 0.66, "color": "blue", "marker": "^"},
    "E5-G4": {"x": 0.53, "y": 0.56, "color": "blue", "marker": "o"},
    "E7-G9": {"x": 0.15, "y": 0.08, "color": "red",  "marker": "^"},
    "E7-G4": {"x": 0.09, "y": 0.06, "color": "red",  "marker": "o"},
}

# =========================
# 2. Figure
# =========================
fig, ax = plt.subplots(figsize=(6, 6), dpi=300)   # 图变小一点

# 1:1 line
ax.plot(
    [0, 0.9], [0, 0.9],
    linestyle="--",
    color="gray",
    linewidth=2,
    label="1:1 line"
)

# =========================
# 3. Scatter points
# =========================
for label, d in data.items():
    ax.scatter(
        d["x"], d["y"],
        s=260,                     # 点更大一点（原来180）
        color=d["color"],
        marker=d["marker"],
        edgecolor="black",
        linewidth=1.2,
        zorder=3
    )

    # label position
    if label == "E7-G4":
        offset = (10, -2)
    elif label == "E7-G9":
        offset = (10, 6)
    elif label == "E5-G4":
        offset = (12, -2)
    else:  # E5-G9
        offset = (10, 0)

    ax.annotate(
        label,
        (d["x"], d["y"]),
        xytext=offset,
        textcoords="offset points",
        fontsize=12,
        fontweight="bold"
    )

# =========================
# 4. Axes, labels, title
# =========================
ax.set_xlim(0, 0.9)
ax.set_ylim(0, 0.9)
ax.set_aspect("equal", adjustable="box")

# tick间隔设为0.2
ticks = np.arange(0, 1.0, 0.2)
ax.set_xticks(ticks)
ax.set_yticks(ticks)

ax.set_xlabel("Conductivity from Pressure Change (mD·ft)", fontsize=13)
ax.set_ylabel("Conductivity from Strain Change (mD·ft)", fontsize=13)
ax.set_title("Pressure- vs. Strain-Derived Conductivity", fontsize=16, fontweight="bold", pad=12)

ax.tick_params(axis='both', labelsize=11)
ax.grid(True, linestyle="--", linewidth=0.8, alpha=0.5)

# =========================
# 5. Legend
# =========================
legend_handles = [
    Line2D([], [], linestyle='none', label='Gauge (shape)'),
    Line2D([], [], marker='o', color='black', linestyle='none', markersize=10, label='Gauge 4'),
    Line2D([], [], marker='^', color='black', linestyle='none', markersize=10, label='Gauge 9'),

    Line2D([], [], linestyle='none', label=''),
    Line2D([], [], linestyle='none', label='Evo (color)'),
    Line2D([], [], marker='o', color='blue', linestyle='none', markersize=10, label='Evo 5'),
    Line2D([], [], marker='o', color='red', linestyle='none', markersize=10, label='Evo 7'),

    Line2D([], [], linestyle='--', color='gray', linewidth=2, label='1:1 line')
]

legend = ax.legend(
    handles=legend_handles,
    loc="upper left",
    frameon=True,
    fontsize=11,
    borderpad=0.8,
    handlelength=2.2
)

# 让两个小标题加粗
for txt in legend.get_texts():
    if txt.get_text() in ["Gauge (shape)", "Evo (color)"]:
        txt.set_fontweight("bold")

# =========================
# 6. Show / Save
# =========================
plt.tight_layout()

plt.show()
```
